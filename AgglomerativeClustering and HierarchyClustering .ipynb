```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
import scipy.cluster.hierarchy as sch

# Load dataset
df = pd.read_csv('/content/drive/MyDrive/DATASETS/sample_-_superstore.csv')

# Select numerical columns
X = df[['Sales', 'Quantity', 'Discount', 'Profit']]

# Remove missing values if any
X = X.dropna()

# Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Agglomerative Clustering
hc = AgglomerativeClustering(n_clusters=3)
labels = hc.fit_predict(X_scaled)

# Cluster Visualization
plt.figure(figsize=(8,5))

plt.scatter(X_scaled[labels==0,0],
            X_scaled[labels==0,3],
            label='Cluster 1')

plt.scatter(X_scaled[labels==1,0],
            X_scaled[labels==1,3],
            label='Cluster 2')

plt.scatter(X_scaled[labels==2,0],
            X_scaled[labels==2,3],
            label='Cluster 3')

plt.xlabel('Sales (Scaled)')
plt.ylabel('Profit (Scaled)')
plt.title('Hierarchical Clustering - Superstore')
plt.legend()
plt.show()

# Dendrogram
plt.figure(figsize=(10,5))
sch.dendrogram(
    sch.linkage(X_scaled, method='ward')
)
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distance')
plt.show()
```
