```python
import transformers
print(transformers.__version__)
```

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I love machine learning!")

print(result)
```

```python
from transformers import pipeline

classifier = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print(classifier("This movie is amazing"))
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)

result = generator(
    "Artificial Intelligence is",
    max_new_tokens=30
)

print(result[0]["generated_text"])
```

```python
from transformers import pipeline

qa = pipeline("question-answering")

result = qa(
    question="What is AI?",
    context="Artificial Intelligence is a field of computer science."
)

print(result)
```

```python
from transformers import pipeline

unmasker = pipeline("fill-mask")

result = unmasker(
    "Machine learning is <mask>."
)

print(result[:3])
```

```python
from transformers import pipeline

summarizer = pipeline("summarization")

text = """
Artificial Intelligence is a branch of computer science that focuses on creating
intelligent machines capable of performing tasks that normally require human intelligence.
"""

result = summarizer(
    text,
    max_length=30,
    min_length=10
)

print(result[0]["summary_text"])
```

```python
from transformers import pipeline

translator = pipeline(
    "text2text-generation",
    model="google/flan-t5-small"
)

result = translator(
    "Translate English to French: Hello, how are you?",
    max_new_tokens=30
)

print(result[0]["generated_text"])
```

```python
from transformers import pipeline

ner = pipeline("ner")

result = ner(
    "Elon Musk founded SpaceX in the United States."
)

print(result)
```

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")

result = classifier(
    "I want to learn deep learning",
    candidate_labels=[
        "technology",
        "sports",
        "politics"
    ]
)

print(result)
```

```python
from transformers import pipeline

extractor = pipeline("feature-extraction")

result = extractor(
    "Machine learning"
)

print(len(result))
```
